```python
# Fill missing Discount_Percent using existing sales values

sales_df["Discount_Percent"] = sales_df["Discount_Percent"].fillna(
    (sales_df["Discount_Amount"] / sales_df["Gross_Sales"]) * 100
)

# Check
print("Missing Discount_Percent after cleaning:",
      sales_df["Discount_Percent"].isnull().sum())
```

```python
# Create Customer_ID -> City mapping
customer_city = customer_df.set_index("Customer_ID")["City"]

# Fill missing City values using Customer_ID
sales_df["City"] = sales_df["City"].fillna(
    sales_df["Customer_ID"].map(customer_city)
)

# Check
print("Missing City after cleaning:",
      sales_df["City"].isnull().sum())
```

```python
sales_numeric = [
    "Quantity",
    "Selling_Price",
    "Cost_Price",
    "Gross_Sales",
    "Discount_Amount",
    "Revenue",
    "Total_Cost",
    "Profit",
    "Profit_Margin_Percent"
]

for col in sales_numeric:
    Q1 = sales_df[col].quantile(0.25)
    Q3 = sales_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((sales_df[col] < lower) | (sales_df[col] > upper)).sum()

    print(f"{col}: {outliers} outliers")
```

```python
# Fill missing Discount_Percent with the median
discount_median = sales_df["Discount_Percent"].median()

sales_df["Discount_Percent"] = sales_df["Discount_Percent"].fillna(discount_median)

print("Missing Discount_Percent after cleaning:",
      sales_df["Discount_Percent"].isnull().sum())

print("Median Discount_Percent used:", discount_median)
```

```python
# Verify sales calculations

sales_df["Calculated_Gross_Sales"] = (
    sales_df["Quantity"] * sales_df["Selling_Price"]
)

sales_df["Calculated_Discount_Amount"] = (
    sales_df["Calculated_Gross_Sales"] *
    sales_df["Discount_Percent"] / 100
)

sales_df["Calculated_Revenue"] = (
    sales_df["Calculated_Gross_Sales"] -
    sales_df["Calculated_Discount_Amount"]
)

print("Gross Sales mismatches:",
      (sales_df["Gross_Sales"] != sales_df["Calculated_Gross_Sales"]).sum())

print("Discount Amount mismatches:",
      (abs(sales_df["Discount_Amount"] -
           sales_df["Calculated_Discount_Amount"]) > 0.01).sum())

print("Revenue mismatches:",
      (abs(sales_df["Revenue"] -
           sales_df["Calculated_Revenue"]) > 0.01).sum())
```

```python
# Verify Profit calculation

sales_df["Calculated_Profit"] = (
    sales_df["Revenue"] - sales_df["Total_Cost"]
)

print(
    "Profit mismatches:",
    (abs(sales_df["Profit"] - sales_df["Calculated_Profit"]) > 0.01).sum()
)
```

```python
# Final validation of important numerical columns

checks = {
    "Negative Quantity": (sales_df["Quantity"] < 0).sum(),
    "Negative Selling Price": (sales_df["Selling_Price"] < 0).sum(),
    "Negative Cost Price": (sales_df["Cost_Price"] < 0).sum(),
    "Invalid Discount (>100%)": (sales_df["Discount_Percent"] > 100).sum(),
    "Negative Revenue": (sales_df["Revenue"] < 0).sum(),
}

for check, count in checks.items():
    print(f"{check}: {count}")
```

```python
print("FINAL MISSING VALUES")

for name, df in {
    "Customer": customer_df,
    "Inventory": inventory_df,
    "Product": product_df,
    "Sales": sales_df
}.items():
    print(f"\n{name}:")
    print(df.isnull().sum()[df.isnull().sum() > 0])
```
